```python
data = [
    (1, "Alice", 2500.00),
    (2, "Bobby", 300.00),
    (3, "cat", 200.00),
    (4, "dog", 100.00)
]

schema = StructType([
    StructField("cust_id", IntegerType(), False),
    StructField("c_name", StringType(), True),
    StructField("c_salary", DoubleType(), True),
    # StructField("c_account", StringType(), True)
])
```

```python
data2 = [
    (5, "checking", 2500.0),
    (None, "null_check", 300.0)
]

df_2 = spark.createDataFrame(data2, schema)
df_2.show()
df_2.printSchema()
```

```python
df = spark.createDataFrame(data, schema)
df.printSchema()
df.show()
```
